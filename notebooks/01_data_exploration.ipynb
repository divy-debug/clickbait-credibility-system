```python
from pathlib import Path

# The notebook is inside /notebooks, so go one level up.
project_root = Path.cwd().parent

print("Project root:")
print(project_root)

print("\nClickbait raw files:")
clickbait_raw = project_root / "data" / "raw" / "clickbait"

for path in clickbait_raw.rglob("*"):
    if path.is_file():
        print(path)

print("\nProcessed files:")
processed = project_root / "data" / "processed"

for path in processed.rglob("*"):
    if path.is_file():
        print(path)
```

```python
import pandas as pd

clickbait_path = project_root / "data" / "processed" / "clickbait.csv"

df = pd.read_csv(clickbait_path)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isna().sum())

print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Class distribution:")
print(df["clickbait_label"].value_counts())

print("\nClass proportions:")
print(df["clickbait_label"].value_counts(normalize=True))
```

```python
print("Shape:", df.shape)

print("\nColumns:")
for column in df.columns:
    print("-", column)

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values:")
print(df.isna().sum())

print("\nDuplicate complete rows:")
print(df.duplicated().sum())

if "id" in df.columns:
    print("\nDuplicate IDs:")
    print(df["id"].duplicated().sum())
```

```python
print("Headline column check:")

if "headline" in df.columns:
    print("Missing headlines:", df["headline"].isna().sum())
    print(
        "Empty headlines:",
        df["headline"].fillna("").str.strip().eq("").sum()
    )
    print("\nHeadline length:")
    print(df["headline"].fillna("").str.len().describe())
else:
    print("ERROR: 'headline' column does not exist.")
```

```python
print("Raw Clickbait-17 files:")

clickbait_raw = project_root / "data" / "raw" / "clickbait"

for path in clickbait_raw.rglob("*"):
    if path.is_file():
        print(path.relative_to(project_root))
```

```python
longest = (
    df.assign(headline_length=df["headline"].str.len())
      .sort_values("headline_length", ascending=False)
      [["id", "headline", "headline_length", "clickbait_score", "clickbait_label"]]
      .head(10)
)

display(longest)
```

```python
for _, row in longest.iterrows():
    print("=" * 100)
    print("ID:", row["id"])
    print("Length:", row["headline_length"])
    print("Score:", row["clickbait_score"])
    print("Label:", row["clickbait_label"])
    print("Headline:")
    print(row["headline"])
```
